# 16.2 Use the csv module and its DictReader method to read books.csv to the variable books. 
## Print the values in books. 

```python
import csv

with open("books.csv", "r", newline="", encoding="utf-8") as file:
    books = list(csv.DictReader(file))

for book in books:
    print(book)

```

## 16.4 Use the sqlite3 module to create a SQLite database called books.db and a table called books 
with these fields: title (text), author (text), and year (integer).


```python
import sqlite3

connection = sqlite3.connect("books.db")
cursor = connection.cursor()

cursor.execute("""
    CREATE TABLE IF NOT EXISTS books (
        title TEXT,
        author TEXT,
        year INTEGER
    )
""")

connection.commit()
connection.close()
```

## 16.5 Read books2.csv and insert its data into the book table.
note: i had to clear previous entries to avoid extraneous entries when 
testing this section


```python
import csv
import sqlite3

connection = sqlite3.connect("books.db")
cursor = connection.cursor()

# Clear previous entries before reloading the assignment's CSV.
cursor.execute("DELETE FROM books")

with open("books2.csv", "r", newline="", encoding="utf-8") as file:
    books = csv.DictReader(file)

    for book in books:
        cursor.execute(
            "INSERT INTO books (title, author, year) VALUES (?, ?, ?)",
            (book["title"], book["author"], int(book["year"]))
        )

connection.commit()
connection.close()

# DictReader uses the CSV header as dictionary keys, so the header is not inserted as a record. 
# The ? placeholders bind each book’s values to the SQL statement, and commit() saves the inserted rows.

```

## 16.6 Select and print the title column from the book table in alphabetical order.


```python
import sqlite3

connection = sqlite3.connect("books.db")
cursor = connection.cursor()


cursor.execute("SELECT title FROM books ORDER BY title ASC")

for row in cursor:
    print(row[0])

connection.close()

```

## 16.7 Select and print all columns from the book table in order of publication.



```python
import sqlite3

connection = sqlite3.connect("books.db")
cursor = connection.cursor()

cursor.execute("SELECT * FROM books ORDER BY year ASC")

for row in cursor:
    print(row)

connection.close()
```

# 16.8: Use SQLAlchemy to print book titles alphabetically.
Note: Had to install SQLAlchemy using PIP


```python
import sqlalchemy as sa

engine = sa.create_engine("sqlite:///books.db")

with engine.connect() as connection:
    result = connection.execute(
        sa.text("SELECT title FROM books ORDER BY title ASC")
    )

    for row in result:
        print(row[0])

engine.dispose()
```
